```python
# Addition of two matrices using list

A = [[1, 2],
     [3, 4]]

B = [[5, 6],
     [7, 8]]

C = [[0, 0],
     [0, 0]]

for i in range(2):
    for j in range(2):
        C[i][j] = A[i][j] + B[i][j]

print("First Matrix:")
print(A)

print("Second Matrix:")
print(B)

print("Addition of matrices:")
print(C)
```

```python
# Addition of two matrices using NumPy

import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

C = A + B

print("First Matrix:")
print(A)

print("Second Matrix:")
print(B)

print("Addition of matrices:")
print(C)
```
