```python
# Student data management using dictionary, tupple and list
students = {
    101: ("Ritesh", "CSE", 85),
    102: ("Rahul", "ECE", 78),
    103: ("Aman", "IT", 90)
}

print("Initial Student Records:")
print(students)
```

```python
#add new student 
students[104] = ("Karan", "CSE", 88)

print("After Adding New Student:")
print(students)
```

```python
students.pop(102)

print("After Deleting Student:")
print(students)
```

```python
students[103] = ("Aman", "CSE", 95)

print("After Updating Student:")
print(students)
```

```python
print("Final Student Records:")

for roll_no, details in students.items():
    print("Roll Number:", roll_no)
    print("Name:", details[0])
    print("Branch:", details[1])
    print("Marks:", details[2])
    print()
```
